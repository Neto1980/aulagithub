```python
# Docstring que é uma string literal usada para documentar módulo, função, classe ou método em Python.
def <nome>(<param_1>, <param_2>, ..., <param_n>):
    ''' Texto documentando sua função...
    '''
  <instruções>
  return resultado
```

```python
# Utilizando o Docstring para documentar a função "media()":
def media(lista: list=[0]) -> float:
  ''' Função para calcular a média de notas passadas por uma lista

  lista: list, default [0]
    Lista com as notas para calcular a média
  return = calculo: float
    Média calculada
  '''
  calculo = sum(lista) / len(lista)
  return calculo
```

```python
# 1ª Atividade: calcular a quantidade de elementos na lista:
lista = [16, 14, 63, 65, 17, 99, 70, 11, 20, 48, 79, 32, 17, 89, 12, 25, 66]
count = 1 # Podemos começar em 1 ou utilizar a função len(lista) para calcular a quantidade de elementos da lista:
maior = lista[0]
menor = lista[0]
for i in lista[1:]:
  count = count + 1
  if i >  maior:
    maior = i
  if i < menor:
    menor = i
soma = sum(lista)
print(f"A lista possui {count} elementos em que o maior deles é {maior} e o menor é {menor}. A soma dos valores presentes na lista é igual a {soma}")
```

```python
# Atividade 3: Criar uma lista que retorne os números multiplos de 3:

lista = [97, 80, 94, 88, 80, 1, 12, 16, 53, 62, 32, 24, 99, 9]

def multiplos_de_3(lista):
  mult = []
  for i in lista:
    if i % 3 == 0:
      mult.append(i)
  return mult
mult_3 = multiplos_de_3(lista)

print(f'Os multiplos de 3 são: {mult_3}')
```

```python
# Atividade 5: Criar um programa para ler 5 notas e eliminar a maior e a menor e tirar a media das 3 que sobrarem:
n1 = float(input('Digite a primeira nota entre 1 e 5: '))
n2 = float(input('Digite a primeira nota entre 1 e 5: '))
n3 = float(input('Digite a primeira nota entre 1 e 5: '))
n4 = float(input('Digite a primeira nota entre 1 e 5: '))
n5 = float(input('Digite a primeira nota entre 1 e 5: '))

notas = [n1, n2, n3, n4, n5]

maior = notas[0]
menor = notas[0]

for i in range(5):
  if notas[i] > maior:
    maior = notas[i]
  if notas[i] < menor:
    menor = notas[i]

media = (sum(notas) - maior - menor) / 3

print(f'A media das notas intermediárias é {media}')


```

```python
# Atividade 6: Você recebeu o desafio de criar um código que calcula os gastos de uma viagem para um das quatro cidades partindo de Recife, sendo elas: Salvador, Fortaleza, Natal e Aracaju:
cidade = input('Digite o nome da cidade que deseja viajar (Salvador / Fotaleza / Natal / Aracaju): ')
dias = int(input('Digite a quantidade de dias: '))
custo = 0

def gasto_hotel(dias):
  gasto_dias = dias * 150
  return gasto_dias

gasto_hotel(dias)

def gasto_gasolina(custo):
  if cidade == 'Salvador':
    custo = (850/14) * 5
    return custo
  elif cidade == 'Fortaleza':
    custo = (800/14) * 5
    return custo
  elif cidade == 'Natal':
    custo = (350/14) * 5
    return custo
  elif cidade == 'Aracaju':
    custo = (550/14) * 5
    return custo

gasto_gasolina(custo)

def gasto_passeio(dias):
  if cidade == 'Salvador':
    gasto = 200 * dias
    return gasto
  elif cidade == 'Fortaleza':
    gasto = 400 * dias
    return gasto
  elif cidade == 'Natal':
    gasto = 250 * dias
    return gasto
  elif cidade == 'Aracaju':
    gasto = 300 * dias
    return gasto

gasto_passeio(dias)

custo_total = gasto_hotel(dias) + gasto_gasolina(custo) + gasto_passeio(dias)

print(f"Com base nos gastos definidos, uma viagem de {dias} dias para {cidade} saindo de Recife custaria R$ {custo_total:.2f} reais")
```

```python
# Atividade 7: Filtrar as palavras que tenham mais de 5 caracteres e armazena-las em uma lista:
# A frase é a seguinte "Aprender Python aqui na Alura é muito bom"

x = input('Entre com uma frase: ')

maior_que_5 = lambda palavras: [p for p in palavras.split() if len(p) > 5]

lista = maior_que_5(x)
print(f'A lista com as palavras maiores que 5 caracteres são:  {lista}')

```
